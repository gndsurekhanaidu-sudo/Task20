```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Priya', 'Kumar', 'Sita'],
    'Age': ['20', '21', '22', '23'],
    'Salary': ['25000.5', '30000.75', '28000.0', '35000.5'],
    'Subscribed': ['Yes', 'No', 'Yes', 'No'],
    'Join_Date': ['2026-01-10', '2026-02-15',
                  '2026-03-20', '2026-04-25']
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.dtypes)
```

```python
df.info()
```

```python
df['Age'] = df['Age'].astype(int)

print(df['Age'])
```

```python
df['Salary'] = pd.to_numeric(df['Salary'])

print(df['Salary'])
```

```python
df['Subscribed'] = df['Subscribed'].map({
    'Yes': True,
    'No': False
})

print(df['Subscribed'])
```

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'])

print(df['Join_Date'])
```

```python
df['Name'] = df['Name'].astype('string')
```

```python
print(df.dtypes)
```

```python
print(df)
```

```python
df.to_csv('cleaned_dataset.csv', index=False)
```

```python
print("DATA TYPE INSPECTION REPORT")

print("\nBefore Conversion:")
print("Age, Salary, Subscribed, Join_Date: object")

print("\nAfter Conversion:")
print(df.dtypes)

print("\nCleaned Dataset:")
print(df)
```
